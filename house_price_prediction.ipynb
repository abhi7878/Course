<h5> 1. Import required libraries<br>
---------------------------------</h5>

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, Lasso, RidgeCV, LassoCV
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.impute import SimpleImputer
```

<h5>2.Load Dataset <br>-----------------------------------</h5>

```python
data_file = "/Users/I338081/Downloads/train.csv"

df = pd.read_csv(data_file)
```

<h5>3. Separate Features & Target <br>-----------------------------------</h5>

```python
X = df.drop("SalePrice", axis=1)
y = df["SalePrice"]
```

```python
df
```

<h5>4.Handle Categorical Variables <br>-----------------------------------</h5>

```python
X = pd.get_dummies(X, drop_first=True)
```

```python
# Remove zero-variance columns
zero_var_cols = X.columns[X.nunique() <= 1]
X = X.drop(columns=zero_var_cols)
```

<h5>5. Train-Test Split <br>-------------------------------------</h5> 

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

<h5>6. Handle Missing Values<br>-----------------------------------</h5>

```python
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)
```

<h5>7. Feature Scaling<br>
-------------------------------------</h5>

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)
```

<h5>8. Ridge Regression (Lambda Optimization)<br>
-------------------------------------------------</h5>

```python
alphas = np.logspace(-3, 3, 100)

ridge_cv = RidgeCV(alphas=alphas, cv=5)
ridge_cv.fit(X_train_scaled, y_train)

best_ridge_alpha = ridge_cv.alpha_

ridge_model = Ridge(alpha=best_ridge_alpha)
ridge_model.fit(X_train_scaled, y_train)

y_test_pred_ridge = ridge_model.predict(X_test_scaled)

ridge_r2 = r2_score(y_test, y_test_pred_ridge)
ridge_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred_ridge))
```

<h5>9. Lasso Regression (Lambda Optimization)<br>
--------------------------------------------------</h5>

```python
lasso_cv = LassoCV(
    alphas=alphas,
    cv=5,
    max_iter=10000,
    random_state=42
)
lasso_cv.fit(X_train_scaled, y_train)

best_lasso_alpha = lasso_cv.alpha_

lasso_model = Lasso(alpha=best_lasso_alpha)
lasso_model.fit(X_train_scaled, y_train)

y_test_pred_lasso = lasso_model.predict(X_test_scaled)

lasso_r2 = r2_score(y_test, y_test_pred_lasso)
lasso_rmse = np.sqrt(mean_squared_error(y_test, y_test_pred_lasso))
```

<h5> 10. Feature Importance <br>
-----------------------------------</h5>

```python
ridge_coeffs = pd.Series(
    ridge_model.coef_, index=X.columns
).sort_values(key=abs, ascending=False)

lasso_coeffs = pd.Series(
    lasso_model.coef_, index=X.columns
)

lasso_selected_features = lasso_coeffs[lasso_coeffs != 0].sort_values(
    key=abs, ascending=False
)
```

<h5>11. Results<br>--------------------------</h5>


```python

print("\n========== RIDGE ==========")
print("Optimal lambda:", best_ridge_alpha)
print("R2:", ridge_r2)
print("RMSE:", ridge_rmse)
print("\nTop Ridge Features:")
print(ridge_coeffs.head(10))

print("\n========== LASSO ==========")
print("Optimal lambda:", best_lasso_alpha)
print("R2:", lasso_r2)
print("RMSE:", lasso_rmse)
print("\nSelected Lasso Features:")
print(lasso_selected_features)
```

<h5> 12. Investment Decision <br>--------------------------------</h5>

```python
X_all_imputed = imputer.transform(X)
X_all_scaled = scaler.transform(X_all_imputed)

df["predicted_price"] = ridge_model.predict(X_all_scaled)

df["decision"] = np.where(
    df["predicted_price"] > df["SalePrice"],
    "BUY",
    "DO NOT BUY"
)

print("\nSample Decisions:")
print(df[["SalePrice", "predicted_price", "decision"]].head())
```
